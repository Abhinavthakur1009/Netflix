```python
url = "https://quotes.toscrape.com"
response = requests.get(url)

soup = BeautifulSoup(response.text, "lxml")

quotes = soup.find_all("span", class_="text")
authors = soup.find_all("small", class_="author")
# print(":"*20)
# print(quotes)
# print(authors)
# print(":"*20)
quote_list = []
author_list = []

for i in range(len(quotes)):
    quote_list.append(quotes[i].text)
    author_list.append(authors[i].text)

print(quote_list[:3])
print(author_list[:3])
```

# 7. Step 4 — Data Cleaning (Important Teaching Part)
    Common cleaning steps:
        Remove extra spaces
        Remove newline characters
        Handle missing values
        Convert data types
    Example:
    clean_quotes = [q.strip() for q in quote_list]

```python
clean_quot_loop = []
for i in quote_list:
    clean_quot_loop.append(i.strip())
clean_quot_loop
```

```python
clean_author_loop = []
for i in author_list:
    clean_author_loop.append(i.strip())
clean_author_loop
```

```python
clean_quotes = [q.strip() for q in quote_list]
clean_author = [au.strip() for au in author_list]
print(":"*50)
print(clean_quotes)
print(":"*50)
print(clean_author)
```

```python
import pandas as pd
data = {
    "Quote": clean_quotes,
    "Author": clean_author
}
print(":")
print(data)
print(":"*50)
df = pd.DataFrame(data)
display(df)
```

```python
data = {
    "Quote": clean_quot_loop,
    "Author": clean_author_loop
}
print(":")
print(data)
print(":"*50)
df = pd.DataFrame(data)
display(df)
```

```python
m_q = len( quote_list[0] )
m_a = len( author_list[0] )
m_qu = quote_list[0] 
m_au = author_list[0]

for i in range(len(quote_list)):
    if len(quote_list[i]) > m_q:
        m_q = len(quote_list[i])
        m_qu = quote_list[i]
        
    if len(author_list[i]) > m_a:
        m_a = len(author_list[i])
        m_au = author_list[i]

print(":"*50)
print("Max length of Qu: ", m_q, m_qu)
print("Max length of au: ", m_a, m_au)
print(":"*50)
```
